# Random Forest Classifier — Heart Disease Classification

## Setup

```python
import sys
from pathlib import Path

# Ensure this notebook's directory (containing utils.py) is importable
sys.path.append(str(Path.cwd()))

from utils import (
    load_data,
    perform_eda,
    preprocess_data,
    evaluate_model,
    save_model,
    append_metrics,
    RANDOM_STATE,
)
```

## Step 1: Load Dataset & Exploratory Data Analysis (EDA)
Displays shape, columns, dtypes, missing values, duplicates, target distribution, and descriptive statistics.

```python
# Step 1: Load dataset and run Exploratory Data Analysis (EDA)
df = load_data()
perform_eda(df)
```

## Step 2: Preprocessing
Missing-value imputation, label/one-hot encoding, feature scaling, and an 80:20 stratified train/test split (random_state=42).

```python
# Step 2: Preprocessing (missing values, encoding, scaling, train/test split)
X_train, X_test, y_train, y_test, scaler, feature_names = preprocess_data(df)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Number of features after encoding: {len(feature_names)}")
```

## Step 3: Train Random Forest

```python
# Step 3: Train Random Forest on the training data
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)
model.fit(X_train, y_train)
print(model)
```

## Step 4: Evaluate Random Forest
Computes Accuracy, ROC AUC, Precision, Recall, F1, and MCC on the held-out test set; prints the confusion matrix and classification report; saves confusion matrix / ROC curve / feature-importance plots.

```python
# Step 4: Evaluate Random Forest on the SAME test set used by every other model
metrics = evaluate_model(
    model=model,
    model_name="Random Forest",
    X_test=X_test,
    y_test=y_test,
    feature_names=feature_names,
)
metrics
```

## Save Trained Model & Record Metrics

```python
# Persist the trained model and append its metrics to the shared
# metrics store, consumed later by model/compare_models.py (Step 5)
save_model(model, "random_forest.joblib")
append_metrics(metrics)
```
